## Syntax for Defining a Function

A function definition has three main parts:

```python
def function_name():
    task_you_want_the_function_to_do
```

What Each Part Means:
- def tells Python that you are defining a function.
- function_name is the name you choose for the function.
- () holds the function's parameters. It is empty when the function does not need any parameters.
- : marks the beginning of the function body.
- The indented line contains the task the function should perform.

## Arguments and Parameters

An **argument** is a concrete value passed to a function when the function is called.

A **parameter** is a placeholder written in the function definition. It receives the argument when the function is called.

### Example

```python
def greet(name):
    print("Hello,", name)
```

name: parameter

```Python
greet("Maya")
```

"Maya": argument

**We discuss several conditions for parameters inside a function**.

## 1. No parameter: print a message

A function can have no parameters. The call below is the test case.

````python
def say_hello():
    print("Hello, Python!")

# Test case: function call
say_hello()
````

## 2. One parameter

The parameter `name` lets the caller provide different input.

````python
def greet(name):
  output = f"Hello, {name}!"
  print(output)

# Test cases
greet("Aisha")
greet("Ben")
````

## 3. Two parameters and one returned value

````python
def add_numbers(first_number, second_number):
  num = first_number + second_number
  return num

# Test case
add_numbers(3, 4)
````

## 4. Return one value

`return` sends a value back to the caller. The caller can store and test it.

````python
def square(number):
  num = number * number
  return num

# Test case
square(5)
````

## 5. Return two values

Multiple returned values are received as a tuple.

````python
def get_name_parts(full_name):
    parts = full_name.split()
    return parts[0], parts[1]

# Test case
first, last = get_name_parts("Maya Chen")
print(first)
print(last)
````

````python
# you can also make function call by only naming one variable
# all the return value more than 2 will be treated as a tuple
name = get_name_parts("Maya Chen")
first = name[0]
last = name[1]
print(first)
print(last)
````

## 6. Return three values

The caller can unpack three returned values into three variables.

````python
def describe_number(number):
    doubled = number * 2
    squared = number * number

    if number > 0:
        category = "positive"
    elif number < 0:
        category = "negative"
    else:
        category = "zero"

    return doubled, squared, category

# Test case
double_value, square_value, category = describe_number(3)
print(double_value, square_value, category)
````

````python
# or call this way, notice all_nums is a tuple
# you can call it just like a list by index
all_nums = describe_number(3)
print(all_nums)
print(all_nums[0])
print(all_nums[1])
print(all_nums[2])
````
